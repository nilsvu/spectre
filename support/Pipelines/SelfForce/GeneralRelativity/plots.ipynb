```python
import h5py
import matplotlib.pyplot as plt
from spectre.Visualization.ReadH5 import to_dataframe
# Load data
filenames = glob.glob("/Users/nilsvu/Projects/spectre/build-Default-Release/test_gsf11/m*/GrSelfForceReductions.h5")
data = []
for filename in filenames:
    m = int(filename.split("m")[-1].split("/")[0])
    with h5py.File(filename, "r") as f:
        df = to_dataframe(f["Redshift.dat"]).set_index("IterationId")
    redshift_re = df["Re(Redshift)"].groupby("IterationId").mean()
    redshift_im = df["Im(Redshift)"].groupby("IterationId").mean()
    data.append({"MMode": m, "Re(Redshift)": redshift_re.iloc[-1], "Im(Redshift)": redshift_im.iloc[-1]})
df = pd.DataFrame(data).set_index("MMode").sort_index()
df /= -np.sqrt(1. - 3.0 / 10.0)
df.iloc[1:] *= 2.0
```
